```python
"""
Q.1 The NumPy Array & Matrix Logic Drill
Focus: Indexing, slicing, boolean masking, and vectorized operations without loops.
"""

import numpy as np

np.random.seed(42)  # remove this line if you want different random values every run

# Task 1: Create a 5x5 matrix with random integers between 1 and 50
matrix = np.random.randint(1, 51, size=(5, 5))
print("Task 1: Original 5x5 Matrix")
print(matrix)

# Task 2: Find and print all numbers greater than 25 using boolean indexing
greater_than_25 = matrix[matrix > 25]
print("\nTask 2: Numbers greater than 25")
print(greater_than_25)

# Task 3: Replace all even numbers with 0 WITHOUT altering the original matrix
# -> work on a copy, since boolean-mask assignment mutates in place
matrix_no_even = matrix.copy()
matrix_no_even[matrix_no_even % 2 == 0] = 0
print("\nTask 3: Matrix with evens replaced by 0 (on a copy)")
print(matrix_no_even)
print("Original matrix is unchanged:")
print(matrix)

# Task 4: Normalize a 1D array of 10 elements so all values fall strictly between 0 and 1
# Formula: (x - min(x)) / (max(x) - min(x))
arr_1d = np.random.randint(1, 100, size=10).astype(float)
normalized = (arr_1d - arr_1d.min()) / (arr_1d.max() - arr_1d.min())
print("\nTask 4: Original 1D array")
print(arr_1d)
print("Normalized array (0 to 1)")
print(normalized)
```

```python
"""
Q.2 The Pandas Data Cleaning & Transformation Challenge
Focus: DataFrames, handling missing values, filtering, and aggregations.
"""

import numpy as np
import pandas as pd

# Task 1: DataFrame with Employee_ID, Name, Department, Age, Salary
# (with a few intentional NaNs in Salary)
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Ravi", "Priya", "Amit", "Sneha", "Karan", "Divya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Age": [25, 30, 28, 35, 40, 29],
    "Salary": [50000, np.nan, 55000, np.nan, 62000, 58000]
}
df = pd.DataFrame(data)
print("Task 1: Original DataFrame")
print(df)

# Task 2: Detect and count missing values in each column
missing_counts = df.isnull().sum()
print("\nTask 2: Missing value counts per column")
print(missing_counts)

# Task 3: Fill missing Salary with the median salary of their respective Department
# groupby + transform lets each NaN get filled with ITS OWN department's median
df["Salary"] = df.groupby("Department")["Salary"].transform(
    lambda x: x.fillna(x.median())
)
print("\nTask 3: DataFrame after filling missing Salary with Department median")
print(df)

# Task 4: Annual_Bonus (10% of Salary) + average salary per department
df["Annual_Bonus"] = df["Salary"] * 0.10
avg_salary_by_dept = df.groupby("Department")["Salary"].mean()
print("\nTask 4: DataFrame with Annual_Bonus column")
print(df)
print("\nAverage Salary per Department")
print(avg_salary_by_dept)
```

```python
"""
Q.3 NumPy Array Reshaping & Slicing Challenge
Focus: Reshaping dimensions and extracting specific sub-grids.
"""

import numpy as np

# Task 1: 1D array containing numbers from 1 to 20 (inclusive)
arr_20 = np.arange(1, 21)
print("Task 1: 1D array (1 to 20)")
print(arr_20)

# Task 2: Reshape into a 4x5 matrix
matrix_4x5 = arr_20.reshape(4, 5)
print("\nTask 2: Reshaped into 4x5 matrix")
print(matrix_4x5)

# Task 3: Extract only the 2nd and 3rd columns (index 1 and 2)
cols_2_3 = matrix_4x5[:, 1:3]
print("\nTask 3: 2nd and 3rd columns")
print(cols_2_3)

# Task 4: Extract the four corner elements into a single 1D array
# Fancy indexing: pairs (0,0), (0,-1), (-1,0), (-1,-1)
rows = np.array([0, 0, -1, -1])
cols = np.array([0, -1, 0, -1])
corners = matrix_4x5[rows, cols]
print("\nTask 4: Four corner elements")
print(corners)
```

```python
"""
Q.4 Pandas Filtering & Conditional Selection
Focus: Complex boolean indexing and conditional updates in DataFrames.
"""

import pandas as pd

# Task 1: DataFrame with Product, Category, Price, and Stock (at least 5 rows)
products = {
    "Product": ["Laptop", "Mouse", "Monitor", "Keyboard", "Webcam"],
    "Category": ["Electronics", "Accessories", "Electronics", "Accessories", "Electronics"],
    "Price": [55000, 450, 8500, 900, 2200],
    "Stock": [12, 150, 40, 18, 60]
}
df_products = pd.DataFrame(products)
print("Task 1: Original DataFrame")
print(df_products)

# Task 2: Filter rows where Price > 500 AND Stock < 50
filtered_products = df_products[(df_products["Price"] > 500) & (df_products["Stock"] < 50)]
print("\nTask 2: Filtered rows (Price > 500 and Stock < 50)")
print(filtered_products)

# Task 3: Add Restock_Needed column (True if Stock < 20, else False)
df_products["Restock_Needed"] = df_products["Stock"] < 20
print("\nTask 3: DataFrame with Restock_Needed column")
print(df_products)

# Task 4: Sort the final DataFrame in descending order by Price
df_sorted = df_products.sort_values(by="Price", ascending=False)
print("\nTask 4: DataFrame sorted by Price (descending)")
print(df_sorted)
```

```python
"""
Q.5 Vectorized Temperature & Math Operations
Focus: High-performance vectorization without Python for loops.
"""

import numpy as np

np.random.seed(42)  # remove this line if you want different random values every run

# Task 1: 10 random floating-point temperatures in Celsius (15.0 to 40.0)
celsius = np.random.uniform(15.0, 40.0, size=10)
print("Task 1: Celsius temperatures")
print(celsius)

# Task 2: Convert all to Fahrenheit simultaneously (vectorized)
# Formula: F = (C * 9/5) + 32
fahrenheit = (celsius * 9 / 5) + 32
print("\nTask 2: Fahrenheit temperatures")
print(fahrenheit)

# Task 3: Square root of a separate 1D array of integers 1 to 10
ints_1_to_10 = np.arange(1, 11)
sqrt_values = np.sqrt(ints_1_to_10)
print("\nTask 3: Square roots of 1-10")
print(sqrt_values)

# Task 4: Index position of the maximum temperature in the Fahrenheit array
max_index = np.argmax(fahrenheit)
print(f"\nTask 4: Index of maximum Fahrenheit temperature -> {max_index} "
      f"(value = {fahrenheit[max_index]:.2f}F)")
```
