**Challenge 1: Number Guessing Limit**

```python
def simulate_guesses(target_num):
    current_guess = 1
    attempts = 0

    while current_guess < target_num:
        attempts += 1
        current_guess += 3


    attempts += 1
    return attempts


target = 25
result = simulate_guesses(target)
print(f"Attempts needed for target {target}: {result}")
```

**Challenge 2: The Budget Alert**

```python
def track_expenses(budget, expense_list):
    total_spent = 0
    for expense in expense_list:
        total_spent += expense
        if total_spent > budget:
            return "Budget Exceeded!"
    return budget - total_spent
my_budget = 100
my_expenses = [20, 45, 40, 10]
result = track_expenses(my_budget, my_expenses)
print(f"Result: {result}")
```

**Challenge 3: The Inventory Tracker**

```python
inventory = ["apple", "banana", "orange"]
def restock_item(new_item):
    if new_item not in inventory:
        inventory.append(new_item)
delivery = ["banana", "grape", "apple", "pear"]
for item in delivery:
    restock_item(item)
print("Final Inventory:", inventory)
```
